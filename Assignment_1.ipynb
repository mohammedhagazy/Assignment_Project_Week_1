```python
#!/usr/bin/env python3
"""
Simple CLI Quiz App (each question = its own function)
--------------------------------------------------------
Every question has its own function. Each function:
  1. Prints the question and its choices
  2. Takes the user's answer with input()
  3. Checks the answer with if/else
  4. Returns True (correct) or False (wrong / invalid input)
 
The main run_quiz() function just calls each question-function once,
in order, and adds up how many returned True.
"""
```

***# ------------------------- Question 1 ------------------------- #***

```python
def ask_question_1():
    print("\nQuestion 1: What is the capital City of France?")
    print("  A) Berlin")
    print("  B) Madrid")
    print("  C) Paris")
    print("  D) Rome")
 
    try:
        answer = input("Your answer (A/B/C/D): ").strip().upper()
    except Exception:
        answer = ""
 
    if answer == "C":
        print("Correct!")
        return True
    else:
        print("Wrong. The correct answer was C) Paris.")
        return False
ask_question_1()
```

***# ------------------------- Question 2 ------------------------- #***

```python
def multiplication_quiz():
    print("Question: What is 12 x 4?")
    print("A) 44")
    print("B) 46")
    print("C) 50")
    print("D) 48")

    answer = input("Choose A, B, C, or D: ").upper()

    if answer == "D":
        print("Correct!")
    elif answer in ["A", "B", "C"]:
        print("Wrong!")
    else:
        print("Invalid choice!")


multiplication_quiz()
```
